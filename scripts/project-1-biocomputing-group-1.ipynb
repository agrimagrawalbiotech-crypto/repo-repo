Retrieving FASTA: https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi?db=nucleotide&id=NM_001093771.3&rettype=fasta&retmode=text

then go to linux command line and copy the file, rename and shift to working directory. 


```python
import requests
base = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi"

def fetch_fasta(accession):
    """Download one nucleotide record as FASTA and save it."""
    params = {"db": "nucleotide", "id": accession,
        "rettype": "fasta", "retmode": "text"}
    r = requests.get(base, params=params)
    if r.status_code != 200:
        return None
    filename = "data/"+ accession + ".fasta"
    with open(filename, "w") as fh:
        fh.write(r.text)
    return filename


print(fetch_fasta("NM_001093771.3"))
```

```python
from Bio import SeqIO 
from Bio.Seq import Seq

TXNRD1_mrna = SeqIO.read("data/" +"NM_001093771.3.fasta", "fasta")
print(TXNRD1_mrna.seq)
print("length of nucleotide sequence: ", len(TXNRD1_mrna.seq))
```
