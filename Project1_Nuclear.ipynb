```python
from Bio import SeqIO # importing Biopython's SeqIO module
```

```python
record = SeqIO.read("sequence (5).fasta", "fasta") # reading the nucleotide FASTA file
print(record.id) # retrieving accession ID
```

Question 2a: 

```python
CDS = record.seq[113:5705] # extracting amino acid sequence from the CDS coordinate
```

```python
length = len(CDS)
print("Length of coding sequence from the mRNA: ",length, "nucleotides")
```

```python
if length % 3 == 0:
    print ("The length of CDS is divisible by 3")
else:
    print ("The length of CDS is not divisible by 3")
```

```python
codon = length / 3 # confirming its divisibility by 3
print("Number of codons after dividing by 3: ",codon)
```

```python
mRNA = CDS.replace("T", "U") # substituting Ts with Us
```

Q.2b

```python
# dictionary of all the codons with single letter amino acid code
codon_table = {
    "UUU": "F",
    "UUC": "F",
    "UUA": "L",
    "UUG": "L",
    "CUU": "L",
    "CUC": "L",
    "CUA": "L",
    "CUG": "L",
    "AUU": "I",
    "AUC": "I",
    "AUA": "I",
    "AUG": "M",
    "GUU": "V",
    "GUC": "V",
    "GUA": "V",
    "GUG": "V",
    "UCU": "S",
    "UCC": "S",
    "UCA": "S",
    "UCG": "S",
    "AGU": "S",
    "AGC": "S",
    "CCU": "P",
    "CCC": "P",
    "CCA": "P",
    "CCG": "P",
    "ACU": "T",
    "ACC": "T",
    "ACA": "T",
    "ACG": "T",
    "GCU": "A",
    "GCC": "A",
    "GCA": "A",
    "GCG": "A",
    "UAU": "Y",
    "UAC": "Y",
    "UAA": "*",
    "UGA": "*",
    "UAG": "*",
    "CAU": "H",
    "CAC": "H",
    "CAA": "Q",
    "CAG": "Q",
    "AAU": "N",
    "AAC": "N",
    "AAA": "K",
    "AAG": "K",
    "GAU": "D",
    "GAC": "D",
    "GAA": "E",
    "GAG": "E",
    "UGU": "C",
    "UGC": "C",
    "UGG": "W",
    "CGU": "R",
    "CGC": "R",
    "CGA": "R",
    "CGG": "R",
    "AGA": "R",
    "AGG": "R",
    "GGU": "G",
    "GGC": "G",
    "GGA": "G",
    "GGG": "G" }
```

```python
#translating all codons to amino acids
i = 0
protein = ""
while i < len(mRNA):
    codons = mRNA[i:i+3]
    translation = codon_table[codons]
    protein += translation
    i += 3
```

```python
print("The first 30 residue of the protein: ", protein[0:30]) # printing first 30 amino acid residues
print("The length of amino acid sequence is:", len(protein), "residues")
```

Q.2c

```python
differences = 0 
record_protein = SeqIO.read("sequence (6).fasta", "fasta") #sequence (6) is the protein FASTA file
for a, b in zip(record_protein.seq, protein):
    if a != b:
        differences += 1
if differences == 0:
    print("The two amino acid sequences are identical.")
else:
    print("The two amino acid sequences are not identical.")
```

Q.2d

```python
function_check_translation = mRNA.translate()
for a, b in zip(function_check_translation, protein):
    if a != b:
        differences += 1
if differences == 0:
    print("The two amino acid sequences are identical.")
else:
    print("The two amino acid sequences are not identical.")
```

```python
length_of_deposited_seq = len(record_protein.seq)
length_of_translated_seq = len(protein)
diff = length_of_translated_seq - length_of_deposited_seq
print(diff)
```
